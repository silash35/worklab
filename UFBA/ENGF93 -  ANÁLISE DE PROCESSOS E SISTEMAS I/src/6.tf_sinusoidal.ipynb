```python
import hickle as hkl
import numpy as np
import sympy as sp

from lib.utils import input_names, output_names

s = sp.Symbol("s")
w = sp.Symbol("w", real=True)

G_matrix = hkl.load("../outputs/G_simplified.hkl")
Gjw_matrix = np.empty(G_matrix.shape, dtype=object)

for i, output_name in enumerate(output_names):
    for j, input_name in enumerate(input_names):
        G = sp.simplify(G_matrix[i, j])
        G_name = f"{output_name}_{input_name}"

        G_jw = sp.simplify(G.subs(s, sp.I * w))
        Gjw_matrix[i, j] = G_jw
```

```python
n_inputs = G_matrix.shape[1]
n_outputs = G_matrix.shape[0]

with open(
    "../outputs/funções_de_transferência_senoidais.txt", "w", encoding="utf-8"
) as f:
    for i in range(n_outputs):
        for j in range(n_inputs):
            f.write(f"G[{i},{j}] =\n")
            G_simplified = Gjw_matrix[i, j].subs(sp.I, sp.Symbol("j"))
            f.write(str(G_simplified))
            f.write("\n\n")


with open(
    "../outputs/funções_de_transferência_senoidais.tex", "w", encoding="utf-8"
) as f:
    for i in range(n_outputs):
        for j in range(n_inputs):
            expr = Gjw_matrix[i, j].subs(sp.I, sp.Symbol("j"))
            f.write(f"G_{{{i + 1},{j + 1}}} = ")
            f.write(sp.latex(expr))
            f.write("\n\n")

hkl.dump(Gjw_matrix, "../outputs/G_jw.hkl")
```
