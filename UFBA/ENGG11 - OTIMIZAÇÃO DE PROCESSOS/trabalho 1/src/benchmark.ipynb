# Configuração do algoritmo genético


```python
import random
import time

import numpy as np
from deap import base, creator, tools

# Configuração do Deep
if not hasattr(creator, "FitnessMin"):
    creator.create("FitnessMin", base.Fitness, weights=(-1.0,))

if not hasattr(creator, "Individual"):
    creator.create("Individual", list, fitness=creator.FitnessMin)  # type: ignore


diversity_threshold = 0.1


def diversity(population):
    """
    Calcula a diversidade da população.
    O objetivo é sinalizar quando parar o algoritmo, afinal quando toda
    a população for muito parecida é sinal que a população convergiu
    e posso interromper o algoritmo.
    """

    if len(population) == 0:
        return 0

    # Converte a população em uma matriz numpy para facilitar o cálculo
    pop_matrix = np.array([ind for ind in population])
    # Calcula a média da população
    mean_ind = np.mean(pop_matrix, axis=0)
    # Calcula a distância euclidiana entre cada indivíduo e a média
    distances = np.linalg.norm(pop_matrix - mean_ind, axis=1)
    # Retorna a soma das distâncias
    return np.sum(distances)


def genetic_algorithm(pop, toolbox, cx_prob, mut_prob, max_gen: int, elitism: int):
    """
    Função customizada para rodar o algoritmo genético.
    Substitui os algoritmos já prontos do Deap, como eaSimple e o eaMuPlusLambda.
    Utiliza convergência como critério de parada.
    """
    fitnesses = map(toolbox.evaluate, pop)
    for ind, fit in zip(pop, fitnesses):
        ind.fitness.values = fit

    n_gen = max_gen
    for current_gen in range(max_gen):
        # Seleciona e clona a próxima geração de indivíduos
        offspring = toolbox.select(pop, len(pop))
        offspring = list(map(toolbox.clone, offspring))

        # Aplica os operadores de crossover e mutação
        for child1, child2 in zip(offspring[::2], offspring[1::2]):
            if random.random() < cx_prob:
                toolbox.mate(child1, child2)
                del child1.fitness.values
                del child2.fitness.values

        for mutant in offspring:
            if random.random() < mut_prob:
                toolbox.mutate(mutant)
                del mutant.fitness.values

        # Reavalia o fitness dos indivíduos modificados
        invalid_ind = [ind for ind in offspring if not ind.fitness.valid]
        fitnesses = map(toolbox.evaluate, invalid_ind)
        for ind, fit in zip(invalid_ind, fitnesses):
            ind.fitness.values = fit

        if elitism > 0:
            # Seleciona elite da geração passada e clona
            elite = list(map(toolbox.clone, tools.selBest(pop, elitism)))
            # Seleciona os melhores da soma entre a elite da geração passada e a nova geração
            pop[:] = tools.selBest(elite + offspring, len(pop))
        else:
            pop[:] = offspring

        if diversity(pop) < diversity_threshold:
            n_gen = current_gen
            break

    return pop, n_gen


# Função para rodar o algoritmo genético
def run_ga(
    bench_function, pop_size: int, cx_prob: float, mut_prob: float, ngen, elitism=0
):
    toolbox = base.Toolbox()
    toolbox.register("rand_float", random.uniform, -5.12, 5.12)
    toolbox.register(
        "individual",
        tools.initRepeat,
        creator.Individual,  # type: ignore
        toolbox.rand_float,  # type: ignore
        n=2,
    )
    toolbox.register("population", tools.initRepeat, list, toolbox.individual)  # type: ignore

    toolbox.register("mate", tools.cxBlend, alpha=0.5)
    toolbox.register("mutate", tools.mutGaussian, mu=0, sigma=1, indpb=0.1)
    toolbox.register("select", tools.selTournament, tournsize=3)
    toolbox.register("evaluate", bench_function)

    pop = toolbox.population(n=pop_size)  # type: ignore

    start_time = time.time()
    final_pop, n_gen = genetic_algorithm(pop, toolbox, cx_prob, mut_prob, ngen, elitism)
    end_time = time.time()

    best_individual = tools.selBest(final_pop, 1)[0]
    best_fitness = toolbox.evaluate(best_individual)[0]  # type: ignore
    return [best_fitness, n_gen, end_time - start_time]
```

# Benchmarks


## Definição de funções úteis


```python
# Fixando seeds
def set_seed(seed: int):
    np.random.seed(seed)
    random.seed(seed)
```

## Função Rastrigin


### Definindo função


```python
import plotly.graph_objects as go


# Definindo a função Rastrigin
def rastrigin(coords, A=10):
    X, Y = coords
    return [
        A * 2 + (X**2 - A * np.cos(2 * np.pi * X)) + (Y**2 - A * np.cos(2 * np.pi * Y))
    ]


# Gera valores de X e Y no intervalo [-5.12, 5.12]
x = np.linspace(-5.12, 5.12, 400)
y = np.linspace(-5.12, 5.12, 400)

X, Y = np.meshgrid(x, y)
Z = rastrigin((X, Y))
Z = np.array(Z).squeeze()

# Criação do gráfico 3D da função Rastrigin
fig = go.Figure(data=[go.Surface(z=Z, x=X, y=Y, colorscale="Viridis")])
fig.update_traces(showscale=False)
fig.update_layout(
    scene=dict(xaxis_title="X", yaxis_title="Y", zaxis_title="Z"),
    autosize=True,
    width=720,
    height=720,
    margin=dict(r=0, l=0, b=0, t=0),
)

zoom = 0.9
fig.update_layout(
    scene_camera=dict(eye=dict(x=1.25 / zoom, y=1.25 / zoom, z=1.25 / zoom))
)
fig.write_image("../media/rastrigin-top.png")

zoom = 1.2
fig.update_layout(scene_camera=dict(eye=dict(x=2 / zoom, y=2 / zoom, z=-1.0 / zoom)))
fig.write_image("../media/rastrigin-bottom.png")

fig.show()
```

### Fazendo benchmark


```python
# Parâmetros do benchmark
pop_size = 20
cx_prob = 0.7
mut_prob = 0.2
ngen = 100

GA_simples = []
GA_elitismo = []


for i in range(10000):
    set_seed(i)
    # Executando GA simples
    GA_simples.append(run_ga(rastrigin, pop_size, cx_prob, mut_prob, ngen))

    set_seed(i)
    # Executando GA com elitismo
    GA_elitismo.append(run_ga(rastrigin, pop_size, cx_prob, mut_prob, ngen, elitism=10))


# Converter valores para numpy
sim = np.array(GA_simples)
elite = np.array(GA_elitismo)

# Extrair as coluna separadamente:
sim_fit = sim[:, 0]
sim_ngen = sim[:, 1]
sim_time = sim[:, 2] * 1e3  # Converter para milissegundos

elite_fit = elite[:, 0]
elite_ngen = elite[:, 1]
elite_time = elite[:, 2] * 1e3  # Converter para milissegundos

print("GA sem Elitismo:")
print(f" - Fitness médio: {np.mean(sim_fit):.2f} ± {np.std(sim_fit):.2f}")
print(f" - N° de gerações médio: {np.mean(sim_ngen):.2f} ± {np.std(sim_ngen):.2f}")
print(f" - Tempo médio/ms: {np.mean(sim_time):.2f} ± {np.std(sim_time):.2f}")

print("GA com Elitismo:")
print(f" - Fitness médio: {np.mean(elite_fit):.2f} ± {np.std(elite_fit):.2f}")
print(f" - N° de gerações médio: {np.mean(elite_ngen):.2f} ± {np.std(elite_ngen):.2f}")
print(f" - Tempo médio/ms: {np.mean(elite_time):.2f} ± {np.std(elite_time):.2f}")
```
